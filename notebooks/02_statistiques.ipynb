# Tests statistiques — retards & satisfaction
Question métier : les retards de livraison dégradent-ils la note des avis ?

```python
import sys

sys.path.insert(0, "../src")
import seaborn as sns

sns.set_theme(style="whitegrid")
from shoppulse.analysis import stats
```

```python
df = stats.load_delivery_satisfaction()
df.describe()
```

## Test 1 — Mann-Whitney U (note : à l'heure vs en retard)
Notes ordinales, distributions non normales → test non paramétrique.

```python
r1 = stats.test_late_vs_score(df)
print(r1.conclusion)
r1
```

## Test 2 — Spearman (durée de livraison vs note)

```python
r2 = stats.test_duration_score_correlation(df)
print(r2.conclusion)
r2
```

## Conclusion
Les deux tests sont hautement significatifs (p ≈ 0). Les retards et les délais longs **dégradent la satisfaction** : cela justifie le modèle prédictif de retard (Lot 4) comme levier de réduction des avis négatifs.
